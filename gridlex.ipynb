```python
class TreeNode:
    def __init__(self,value):
        self.value = value
        self.left =None
        self.right = None


root = TreeNode(10)

root.left = TreeNode(5)
root.right = TreeNode(15)


root.left.left = TreeNode(3)
root.left.right = TreeNode(7)



def preorder(node):
    if node is None:
        return
    print(node.value,end=" ")

    preorder(node.left)
    preorder(node.right)

preorder(root)


def inorder(node):
    if node is None:
        return
    
    inorder(node.left)

    print(node.value,end = " ")

    inorder(node.right)

inorder(root)
```
